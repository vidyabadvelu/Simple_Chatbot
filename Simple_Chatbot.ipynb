```python
# ==========================================
# SIMPLE CHATBOT
# Text and Speech Analysis (TSA)
# Python + Google Colab
# ==========================================

print("====================================")
print("        🤖 TSA CHATBOT")
print("====================================")
print("Type 'bye' to exit the chatbot.\n")

while True:
    user = input("You: ").lower()

    if user == "hello" or user == "hi":
        print("Bot: Hello! How can I help you?")

    elif "how are you" in user:
        print("Bot: I am fine! Thank you for asking.")

    elif "your name" in user:
        print("Bot: My name is TSA Chatbot.")

    elif "what is tsa" in user:
        print("Bot: TSA stands for Text and Speech Analysis.")

    elif "python" in user:
        print("Bot: Python is a popular programming language.")

    elif "thank" in user:
        print("Bot: You're welcome!")

    elif user == "bye":
        print("Bot: Goodbye! Have a nice day.")
        break

    else:
        print("Bot: Sorry, I don't understand that.")
```
